# GTA VI YouTube Live/VOD -> Google Drive Inbox Bridge
Use this 1-click notebook whenever you want to capture a **YouTube-only** live stream (e.g., IShowSpeed, Typical Gamer, Valkyrae) directly into your `GTA6_Inbox` folder on Google Drive. Your GitHub Actions bot will automatically detect the file in Drive, clip it into 9:16 Shorts with word-pop captions, save the finished Shorts to `GTA6_Output`, and schedule them on YouTube Studio!

```
# Step 1: Mount Google Drive & Install yt-dlp + FFmpeg
from google.colab import drive
drive.mount('/content/drive')
!pip install -q -U yt-dlp
!apt-get update -qq && apt-get install -y -qq ffmpeg
import os
INBOX_DIR = '/content/drive/MyDrive/GTA6_Inbox'
os.makedirs(INBOX_DIR, exist_ok=True)
print('Ready! Dropping videos into:', INBOX_DIR)
```

```
# Step 2: Paste any YouTube Live or VOD URL & Streamer Name
STREAMER_NAME = 'IShowSpeed'  #@param {type:'string'}
STREAM_TITLE = 'GTA 6 Vice City Launch Live'  #@param {type:'string'}
YOUTUBE_URL = 'https://www.youtube.com/@IShowSpeed/live'  #@param {type:'string'}
SLICE_MINUTES = 15  #@param {type:'integer'}

import time
out_file = f"{INBOX_DIR}/{STREAMER_NAME} - {STREAM_TITLE} ({int(time.time())}).mp4"
!yt-dlp -f "bv*[height<=720]+ba/b[height<=720]/b" --downloader ffmpeg --downloader-args "ffmpeg_i:-t {SLICE_MINUTES * 60}" --merge-output-format mp4 -o "{out_file}" "{YOUTUBE_URL}"
print('Saved to Google Drive Inbox:', out_file)
```
